```python
import numpy as np
import random
import scipy
import matplotlib.pyplot as plt
random.seed(20020408)

%matplotlib inline
```

## Stimulus history schematic
$f(t)$ as our input

```python
datapoints = 200
t = np.linspace(-100, 0, datapoints)

events = np.zeros(datapoints)
event_time_ls = [40, 104, 130, 172]
events[event_time_ls] = 1

for event_time in event_time_ls:
    rnd_event_end = event_time + random.randint(1,10)
    print(event_time, rnd_event_end)
    events[event_time:rnd_event_end] = 1

spike_times = t[events == 1]

# Publication-style formatting
plt.rcParams.update({
    "font.family": "sans-serif",
    "font.size": 12,
    "axes.linewidth": 1.2,
    "xtick.direction": "out",
})

fig, ax = plt.subplots(figsize=(8,2))

# Draw spikes
ax.step(
    t,
    events,
    where="post",
    linewidth=2.0,
    color="#2563EB"
)

# Axis labels and limits
ax.set_xlabel("Time (ms)", fontsize=13, labelpad=10)
ax.set_xlim(-100, 0)

# Remove unnecessary elements
ax.set_yticks([0])
ax.set_ylabel("Stimulus", fontsize=13, labelpad=12)
ax.set_title('Stimulus history: f(t)')

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.spines["left"].set_visible(False)

ax.tick_params(axis="x", length=5, width=1.2)

plt.tight_layout()

fig.savefig(
    "stimulus_history.pdf",
    bbox_inches="tight"
)

fig.savefig(
    "stimulus_history.png",
    dpi=500,
    bbox_inches="tight"
)
plt.show()
```
